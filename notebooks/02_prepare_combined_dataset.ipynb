```python
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/raw")
OUTPUT_DIR = Path("../data/processed")


OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

categories = ["airplane", "apple", "bicycle", "cat", "house"]

total_images = sum(
    np.load(DATA_DIR / f"{category}.npy", mmap_mode="r").shape[0]
    for category in categories
)

X = np.lib.format.open_memmap(
    OUTPUT_DIR / "images.npy",
    mode = "w+",
    dtype = np.uint8,
    shape = (total_images, 784)
)

y = np.lib.format.open_memmap(
    OUTPUT_DIR / "labels.npy",
    mode = "w+",
    dtype = np.uint8,
    shape = (total_images,)
)

start = 0

for label, category in enumerate(categories):
    data = np.load(DATA_DIR / f"{category}.npy", mmap_mode = "r")
    end = start + len(data)
    X[start:end] = data
    y[start:end] = label
    print(f"{category}: {len(data)} images copied")
    start = end

X.flush()
y.flush()

print("\nDataset preparation complete!")
print(f"Images: {X.shape}")
print(f"Labels: {y.shape}")
print(f"Total images: {total_images}")
    
    
```
