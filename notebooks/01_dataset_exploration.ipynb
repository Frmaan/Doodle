```python
import numpy as np
from pathlib import Path

DATA_DIR = Path("../data/raw")

categories = sorted([
    file.stem
    for file in DATA_DIR.glob("*.npy")
])

print(f"Total categories: {len(categories)}")

for category in categories:
    file_path = DATA_DIR / f"{category}.npy"
    data = np.load(file_path, mmap_mode="r")
    print(f"\nCategory: {category}")
    print(f"Number of drawings: {len(data):,}")
    print(f"Shape: {data.shape}")
    print(f"Data Type: {data.dtype}")
    print(f"Minimum pixel value: {data.min()}")
    print(f"Maximum pixel value: {data.max()}")
```

```python
# visualize some drawings

import matplotlib.pyplot as plt

images_per_category = 5

fig, axes = plt.subplots(
    len(categories),
    images_per_category,
    figsize=(12, len(categories) * 2)
)

for row, category in enumerate(categories):

    data = np.load(DATA_DIR / f"{category}.npy", mmap_mode="r")

    np.random.seed(42)

    for col in range(images_per_category):

        idx = np.random.randint(0, len(data))
        image = data[idx].reshape(28, 28)

        axes[row, col].imshow(image, cmap="gray")
        axes[row, col].axis("off")

        if col == 0:
            axes[row, col].set_ylabel(
                category,
                rotation=0,
                labelpad=45,
                fontsize=9
            )

plt.tight_layout()
plt.show()
```
