```python
import numpy as np
from pathlib import Path
from sklearn.model_selection import train_test_split

DATA_DIR = Path("../data/processed")

X = np.load(DATA_DIR / "images.npy", mmap_mode="r")
y = np.load(DATA_DIR / "labels.npy", mmap_mode="r")

categories = ["airplane", "apple", "bicycle", "cat", "house"]

print(f"Images: {X.shape}")
print(f"Labels: {y.shape}")
```

```python
indices = np.arange(len(y))

train_idx, temp_idx = train_test_split(
    indices,
    test_size = 0.30,
    random_state = 42,
    stratify = y
)

val_idx, test_idx = train_test_split(
    temp_idx,
    test_size = 0.50,
    random_state = 42,
    stratify = y[temp_idx]
)

print(f"Training samples: {len(train_idx)}")
print(f"Validation samples: {len(val_idx)}")
print(f"Testing samples: {len(test_idx)}")
    
```

```python
# verify class distribution

splits = {
    "Training": train_idx,
    "Validation": val_idx,
    "Testing": test_idx
}

for split_name, split_indices in splits.items():

    print(f"\n{split_name}:")

    labels, counts = np.unique(
        y[split_indices],
        return_counts=True
    )

    for label, count in zip(labels, counts):
        print(f"{categories[label]}: {count}")
```

```python
# save the split indices

np.savez(
    DATA_DIR / "split_indices.npz",
    train_idx = train_idx,
    val_idx = val_idx,
    test_idx = test_idx
)
```
